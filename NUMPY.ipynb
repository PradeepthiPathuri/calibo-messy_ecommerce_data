```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

df = pd.read_csv("house_prices_500.csv")

df = df.drop_duplicates()

for col in ["Area", "Bedrooms", "Bathrooms", "Age", "Price"]:
    df[col] = df[col].fillna(df[col].median())

df["Location"] = df["Location"].fillna(df["Location"].mode()[0])

most_expensive = df.loc[df["Price"].idxmax()]
cheapest = df.loc[df["Price"].idxmin()]

bedroom_price = df.groupby("Bedrooms")["Price"].mean()
bathroom_price = df.groupby("Bathrooms")["Price"].mean()
location_price = df.groupby("Location")["Price"].mean().sort_values()

correlation = df[
    ["Area", "Bedrooms", "Bathrooms", "Age", "Price"]
].corr()

locations = df["Location"].unique()

fig, ax = plt.subplots(2, 4, figsize=(20, 10))

fig.suptitle(
    "Housing Market Analysis Dashboard (500 Properties)",
    fontsize=20,
    fontweight="bold"
)

ax[0, 0].hist(
    df["Price"],
    bins=15,
    color="skyblue",
    edgecolor="black"
)

ax[0, 0].set_title("1. Price Distribution", fontweight="bold")
ax[0, 0].set_xlabel("Price (₹ in Millions)")
ax[0, 0].set_ylabel("Frequency")
ax[0, 0].legend(["Properties"])
ax[0, 0].grid(True, alpha=0.3)


ax[0, 1].scatter(
    df["Area"],
    df["Price"],
    color="tomato",
    edgecolor="black",
    alpha=0.7,
    label="Homes"
)

ax[0, 1].set_title("2. Area vs Price", fontweight="bold")
ax[0, 1].set_xlabel("Area (sqft)")
ax[0, 1].set_ylabel("Price (₹ in Millions)")
ax[0, 1].legend()
ax[0, 1].grid(True, alpha=0.3)

ax[0, 1].annotate(
    "Peak Price",
    xy=(
        most_expensive["Area"],
        most_expensive["Price"]
    ),
    xytext=(
        most_expensive["Area"] - 900,
        most_expensive["Price"] - 1
    ),
    arrowprops=dict(arrowstyle="->")
)


ax[0, 2].bar(
    bedroom_price.index.astype(str),
    bedroom_price.values,
    color="mediumseagreen",
    edgecolor="black",
    label="Avg Price"
)

ax[0, 2].set_title("3. Bedrooms vs Avg Price", fontweight="bold")
ax[0, 2].set_xlabel("Bedrooms")
ax[0, 2].set_ylabel("Avg Price (₹ Millions)")
ax[0, 2].legend()
ax[0, 2].grid(axis="y", alpha=0.3)


ax[0, 3].bar(
    bathroom_price.index.astype(str),
    bathroom_price.values,
    color="orchid",
    edgecolor="black",
    label="Avg Price"
)

ax[0, 3].set_title("4. Bathrooms vs Avg Price", fontweight="bold")
ax[0, 3].set_xlabel("Bathrooms")
ax[0, 3].set_ylabel("Avg Price (₹ Millions)")
ax[0, 3].legend()
ax[0, 3].grid(axis="y", alpha=0.3)


ax[1, 0].scatter(
    df["Age"],
    df["Price"],
    color="gold",
    edgecolor="black",
    alpha=0.7,
    label="Homes"
)

ax[1, 0].set_title("5. Property Age vs Price", fontweight="bold")
ax[1, 0].set_xlabel("Age (Years)")
ax[1, 0].set_ylabel("Price (₹ in Millions)")
ax[1, 0].legend()
ax[1, 0].grid(True, alpha=0.3)

ax[1, 0].annotate(
    "Cheapest",
    xy=(
        cheapest["Age"],
        cheapest["Price"]
    ),
    xytext=(
        cheapest["Age"] + 2,
        cheapest["Price"] + 2
    ),
    arrowprops=dict(arrowstyle="->")
)


ax[1, 1].bar(
    location_price.index,
    location_price.values,
    color="coral",
    edgecolor="black",
    label="Avg Price"
)

ax[1, 1].set_title("6. Location vs Avg Price", fontweight="bold")
ax[1, 1].set_xlabel("Location")
ax[1, 1].set_ylabel("Avg Price (₹ Millions)")
ax[1, 1].tick_params(axis="x", rotation=35)
ax[1, 1].legend()
ax[1, 1].grid(axis="y", alpha=0.3)


price_by_location = [
    df[df["Location"] == location]["Price"]
    for location in locations
]

ax[1, 2].boxplot(
    price_by_location,
    labels=locations,
    patch_artist=True,
    boxprops=dict(facecolor="lightsteelblue")
)

ax[1, 2].set_title("7. Price Box Plot by Location", fontweight="bold")
ax[1, 2].set_xlabel("Location")
ax[1, 2].set_ylabel("Price (₹ in Millions)")
ax[1, 2].tick_params(axis="x", rotation=35)
ax[1, 2].grid(axis="y", alpha=0.3)


heatmap = ax[1, 3].imshow(
    correlation,
    cmap="viridis",
    aspect="auto"
)

fig.colorbar(
    heatmap,
    ax=ax[1, 3],
    fraction=0.046,
    pad=0.04
)

ax[1, 3].set_title(
    "8. Feature Correlation Heatmap",
    fontweight="bold"
)

ax[1, 3].set_xticks(range(len(correlation.columns)))
ax[1, 3].set_yticks(range(len(correlation.columns)))

ax[1, 3].set_xticklabels(
    correlation.columns,
    rotation=45
)

ax[1, 3].set_yticklabels(
    correlation.columns
)

for i in range(len(correlation.columns)):
    for j in range(len(correlation.columns)):
        ax[1, 3].text(
            j,
            i,
            f"{correlation.iloc[i, j]:.2f}",
            ha="center",
            va="center",
            color="white",
            fontweight="bold"
        )

plt.tight_layout(rect=[0, 0, 1, 0.95])

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("online_store_sales.csv")

df["Date"] = pd.to_datetime(df["Date"])

print("First 5 Rows:")
print(df.head())

print("\nDataset Shape:")
print(df.shape)

print("\nDataset Information:")
df.info()

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Records:")
print(df.duplicated().sum())

monthly_sales = df.groupby(
    df["Date"].dt.to_period("M")
)["Sales"].sum()

monthly_sales.index = monthly_sales.index.astype(str)

top_products = (
    df.groupby("Product")["Sales"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

region_sales = (
    df.groupby("Region")["Sales"]
    .sum()
    .sort_values(ascending=False)
)

category_sales = (
    df.groupby("Category")["Sales"]
    .sum()
    .sort_values(ascending=False)
)

highest_month = monthly_sales.idxmax()
highest_sales = monthly_sales.max()

best_product = top_products.idxmax()
best_product_sales = top_products.max()

best_region = region_sales.idxmax()
best_region_sales = region_sales.max()

best_category = category_sales.idxmax()
best_category_sales = category_sales.max()

print("\nHighest Sales Month:")
print(highest_month, "₹", highest_sales)

print("\nTop Product:")
print(best_product, "₹", best_product_sales)

print("\nBest Region:")
print(best_region, "₹", best_region_sales)

print("\nBest Category:")
print(best_category, "₹", best_category_sales)

fig, ax = plt.subplots(2, 2, figsize=(16, 10))

fig.suptitle(
    "E-Commerce Sales Dashboard",
    fontsize=20,
    fontweight="bold"
)

ax[0, 0].plot(
    monthly_sales.index,
    monthly_sales.values,
    marker="o",
    linewidth=2,
    color="royalblue",
    label="Monthly Sales"
)

ax[0, 0].set_title(
    "1. Monthly Sales",
    fontweight="bold"
)

ax[0, 0].set_xlabel("Month")
ax[0, 0].set_ylabel("Sales (₹)")
ax[0, 0].tick_params(
    axis="x",
    rotation=45
)
ax[0, 0].legend()
ax[0, 0].grid(
    True,
    alpha=0.3
)

ax[0, 0].annotate(
    "Highest Sales",
    xy=(
        highest_month,
        highest_sales
    ),
    xytext=(
        highest_month,
        highest_sales * 1.15
    ),
    arrowprops=dict(
        arrowstyle="->"
    )
)

ax[0, 1].bar(
    top_products.index,
    top_products.values,
    color="mediumseagreen",
    edgecolor="black",
    label="Sales"
)

ax[0, 1].set_title(
    "2. Top 10 Products by Sales",
    fontweight="bold"
)

ax[0, 1].set_xlabel("Product")
ax[0, 1].set_ylabel("Sales (₹)")
ax[0, 1].tick_params(
    axis="x",
    rotation=45
)
ax[0, 1].legend()
ax[0, 1].grid(
    axis="y",
    alpha=0.3
)

ax[1, 0].bar(
    region_sales.index,
    region_sales.values,
    color="darkorange",
    edgecolor="black",
    label="Sales"
)

ax[1, 0].set_title(
    "3. Sales Across Regions",
    fontweight="bold"
)

ax[1, 0].set_xlabel("Region")
ax[1, 0].set_ylabel("Sales (₹)")
ax[1, 0].legend()
ax[1, 0].grid(
    axis="y",
    alpha=0.3
)

ax[1, 1].pie(
    category_sales.values,
    labels=category_sales.index,
    autopct="%1.1f%%",
    startangle=90
)

ax[1, 1].set_title(
    "4. Category-wise Sales",
    fontweight="bold"
)

plt.tight_layout(
    rect=[0, 0, 1, 0.95]
)

plt.show()

plt.figure(figsize=(8, 5))

plt.scatter(
    df["Quantity"],
    df["Sales"],
    color="crimson",
    alpha=0.7,
    edgecolor="black"
)

plt.title(
    "Quantity Sold vs Revenue",
    fontweight="bold"
)

plt.xlabel("Quantity Sold")
plt.ylabel("Revenue (₹)")
plt.grid(
    True,
    alpha=0.3
)

plt.show()

print("\n" + "=" * 50)
print("FINAL CONCLUSION")
print("=" * 50)

print(
    f"The best-performing product category is {best_category}, "
    f"with total sales of ₹{best_category_sales:,.0f}."
)

print(
    f"The best-performing region is {best_region}, "
    f"with total sales of ₹{best_region_sales:,.0f}."
)

print(
    f"The highest-selling month is {highest_month}, "
    f"with sales of ₹{highest_sales:,.0f}."
)

print(
    f"The top product is {best_product}, "
    f"with sales of ₹{best_product_sales:,.0f}."
)
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv("covid_data.csv")

df["Date"] = pd.to_datetime(df["Date"])

print(df.head())
print(df.info())

sns.set_theme(style="whitegrid")


plt.figure(figsize=(12, 6))

sns.lineplot(
    data=df,
    x="Date",
    y="Confirmed",
    hue="Country",
    marker="o"
)

plt.title("Daily Confirmed COVID-19 Cases")
plt.xlabel("Date")
plt.ylabel("Confirmed Cases")
plt.xticks(rotation=45)
plt.show()



country_total = df.groupby("Country")["Confirmed"].max().reset_index()

plt.figure(figsize=(10, 6))

sns.barplot(
    data=country_total,
    x="Country",
    y="Confirmed"
)

plt.title("Maximum Confirmed Cases by Country")
plt.xlabel("Country")
plt.ylabel("Confirmed Cases")
plt.xticks(rotation=45)
plt.show()



plt.figure(figsize=(12, 6))

sns.lineplot(
    data=df,
    x="Date",
    y="Deaths",
    hue="Country",
    marker="o"
)

plt.title("COVID-19 Deaths Over Time")
plt.xlabel("Date")
plt.ylabel("Deaths")
plt.xticks(rotation=45)
plt.show()



plt.figure(figsize=(12, 6))

sns.lineplot(
    data=df,
    x="Date",
    y="Recovered",
    hue="Country",
    marker="o"
)

plt.title("COVID-19 Recoveries Over Time")
plt.xlabel("Date")
plt.ylabel("Recovered Cases")
plt.xticks(rotation=45)
plt.show()



heatmap_data = df.groupby("Country")[
    ["Confirmed", "Deaths", "Recovered", "Active"]
].max()

plt.figure(figsize=(10, 7))

sns.heatmap(
    heatmap_data,
    annot=True,
    fmt="g",
    cmap="Reds",
    linewidths=0.5
)

plt.title("COVID-19 Country-wise Statistics")
plt.xlabel("COVID-19 Statistics")
plt.ylabel("Country")
plt.show()



correlation = df[
    ["Confirmed", "Deaths", "Recovered", "Active"]
].corr()

plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    linewidths=0.5
)

plt.title("Correlation Between COVID-19 Variables")
plt.show()



growth = df.groupby("Country").agg(
    Start_Confirmed=("Confirmed", "first"),
    End_Confirmed=("Confirmed", "last")
).reset_index()

growth["Growth"] = (
    growth["End_Confirmed"] - growth["Start_Confirmed"]
)

growth = growth.sort_values(
    by="Growth",
    ascending=False
)

print("\nCountries with Highest COVID-19 Growth:")
print(growth)
```
